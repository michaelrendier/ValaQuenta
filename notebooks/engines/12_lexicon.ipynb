The lexicon is the **inverse** of a dictionary. A dictionary maps a string to a
meaning. This maps a prime — a Riemann zero — to the set of surface forms
observed pointing at it.

> The prime preexists every alphabet invented to point at it.

**Status: untested at scale.** No binary corpus has been loaded. This notebook
exercises the API in memory; it does not demonstrate the engine at archive
scale, and the wiki records that as open.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../../..'))
import math, cmath
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 10,
                     'axes.spines.top': False, 'axes.spines.right': False})
print('python', sys.version.split()[0])
```

```python
from ValaQuenta import Lexicon
from ValaQuenta import RIEMANN_ZEROS

L = Lexicon()
gamma = RIEMANN_ZEROS[0]
print(f'using gamma_1 = {gamma!r}')

# One prime, many faces -- the same word in several languages.
for surface, dom in [('tree', 'botany'), ('alda', 'Quenya'),
                     ('arbre', 'French'), ('Baum', 'German'),
                     ('tree', 'graph theory'), ('tree', 'botany')]:
    L.record(gamma, surface, domain=dom)

print(f'face_count({gamma:.4f}) = {L.face_count(gamma)!r}')
print('faces:', L.faces(gamma))
print('best_face          :', L.best_face(gamma))
print('best_face (Quenya) :', L.best_face(gamma, language_hint="Quenya"))
```

## Several primes

```python
for g, words in [(RIEMANN_ZEROS[1], ['river', 'sirya', 'fleuve']),
                 (RIEMANN_ZEROS[2], ['stone', 'sar', 'pierre'])]:
    for w in words:
        L.record(g, w)

print('known gammas:')
for g in L.known_gammas():
    print(f'  {g!r:>22}  faces={L.face_count(g)}  best={L.best_face(g)!r}')
```

## Persistence and merge

```python
import tempfile, os as _os
path = _os.path.join(tempfile.mkdtemp(), 'lexicon.json')
L.save(path) if hasattr(L, 'save') else None
print('save attempted ->', path, 'exists:', _os.path.exists(path))

L2 = Lexicon()
L2.record(RIEMANN_ZEROS[0], 'arbor', domain='Latin')
L.merge(L2)
print(f'after merge, faces at gamma_1: {L.faces(RIEMANN_ZEROS[0])}')
print()
print('already_seen(some/path) =', L.already_seen('some/path'))
```
