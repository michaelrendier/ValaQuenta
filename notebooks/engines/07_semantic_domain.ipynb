The description says where meaning starts and where it stops. The domain does
not invent the instruments — the music chooses them.

The governing relation is `tau * T_H = constant`: coherence time against Hawking
temperature. A narrow domain is hot and short-lived; a wide one is cold and
persistent.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../../..'))
import math, cmath
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 10,
                     'axes.spines.top': False, 'axes.spines.right': False})
print('python', sys.version.split()[0])
```

```python
from ValaQuenta import SemanticDomain

d = SemanticDomain('the Riemann zeros on the critical line',
                   gamma_min=14.0, gamma_max=50.0)
print('domain      :', d)
print(f'  span      = {d.span!r}')
print(f'  T_Hawking = {d.hawking_temperature()!r}')
print(f'  contains(21.02) = {d.contains(21.02)!r}')
print(f'  contains(99.9)  = {d.contains(99.9)!r}')
```

## tau * T_H is constant across domain widths

```python
from ValaQuenta import RIEMANN_ZEROS
zeros = list(RIEMANN_ZEROS[:20])

print(f'{"span":>9} | {"T_H":>22} | {"coherence tau":>22} | {"tau * T_H":>22}')
print('-' * 84)
rows = []
for width in [5.0, 10.0, 20.0, 40.0, 80.0, 160.0]:
    dd = SemanticDomain('probe', gamma_min=14.0, gamma_max=14.0 + width)
    T = dd.hawking_temperature()
    tau = dd.coherence_time(zeros)
    rows.append((width, T, tau, tau * T))
    print(f'{width:9.1f} | {T!r:>22} | {tau!r:>22} | {tau*T!r:>22}')

prods = [r[3] for r in rows]
print()
print(f'product min = {min(prods)!r}')
print(f'product max = {max(prods)!r}')
print(f'spread      = {(max(prods)-min(prods)):.3e}')
```

```python
ws = [r[0] for r in rows]
fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.4))
a1.loglog(ws, [r[1] for r in rows], 'o-', color='#b04a3f', label='T_H')
a1.loglog(ws, [r[2] for r in rows], 's-', color='#3f7fb0', label='tau')
a1.set_xlabel('domain span'); a1.legend(fontsize=8)
a1.set_title('temperature and coherence time')
a2.semilogx(ws, [r[3] for r in rows], 'o-', color='#2f2f2f')
a2.set_xlabel('domain span'); a2.set_ylabel('tau * T_H')
a2.set_title('the product')
plt.tight_layout(); plt.show()
```

## Instruments: the music chooses them

```python
d2 = SemanticDomain('a narrow band', gamma_min=14.0, gamma_max=26.0)
inst = d2.instruments(zeros)
print(f'domain span {d2.span!r} selects {len(inst)} instruments from {len(zeros)} zeros:')
print(' ', inst)
print()
print(f'is_collapsed = {d2.is_collapsed(zeros)!r}')
print(f'snap(t=1.0)  = {d2.snap(1.0, zeros)!r}')
```
