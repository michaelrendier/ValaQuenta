```python
from ValaQuenta import SemanticWord

w = SemanticWord('tree')
print('word        :', w)
print(f'  surface   = {w.surface!r}')
print(f'  prime     = {w.prime!r}')
print(f'  Re(prime) = {w.prime.real!r}   <- always 1/2')
print(f'  gamma     = {w.gamma!r}')
print(f'  magnitude = {w.magnitude!r}')
print(f'  observer  = {w.observer!r}')
print(f'  faces     = {w.faces()!r}')
```

## Projections: many faces, one prime

```python
w2 = SemanticWord('tree')
for context, value in [('botany', 'woody perennial plant'),
                       ('graph theory', 'acyclic connected graph'),
                       ('computing', 'hierarchical data structure'),
                       ('Quenya', 'alda')]:
    w2.add_projection(context, value)

print(f'faces = {w2.faces()}')
for k, v in w2.projections.items():
    print(f'  {k:>14} -> {v}')
print()
print('Four faces. One prime. The prime preexists every alphabet invented')
print('to point at it.')
print(f'Re(prime) unchanged by adding faces: {w2.prime.real!r}')
```

## Re(prime) = 1/2 holds across every surface form
