The galactic bar swings between two phases:

- **contracted** — the Witches Hat, a 1-arm bar at the outer Lindblad resonance
- **expanded** — the brim, a 2-arm spiral at the inner Lindblad resonance

The contracted half-angle has `tan = d*` exactly. The expanded half-angle is
THE ANGLE, pi/8, from the zero lattice. The bell swings between the spectral
floor and the zero-divisor angle, and **never closes**.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../../..'))
import math, cmath
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 10,
                     'axes.spines.top': False, 'axes.spines.right': False})
print('python', sys.version.split()[0])
```

```python
from ValaQuenta import telperion as tp

print(f'D_STAR          = {tp.D_STAR!r}')
print(f'GAP             = {tp.GAP!r}')
print(f'BELL_AMPLITUDE  = {tp.BELL_AMPLITUDE!r}')
print(f'BRIM_HALF_ANGLE = {tp.BRIM_HALF_ANGLE!r} rad = '
      f'{math.degrees(tp.BRIM_HALF_ANGLE)!r} deg')

bell = tp.bell_geometry()
for k, v in bell.items():
    if isinstance(v, float):
        print(f'  {k:>26}: {v!r}')
    else:
        print(f'  {k:>26}: {v}')
```

## The two anchors of the swing

`tan(contracted) = d* = 0.246` and `tan(expanded) = sqrt(2)-1`. These are not
fitted to a galaxy; they are the spectral floor and THE ANGLE.

```python
print(f'tan(contracted) = {bell["tan_contracted"]!r}')
print(f'D_STAR          = {tp.D_STAR!r}')
print(f'  equal         : {bell["tan_contracted"] == tp.D_STAR}')
print()
print(f'tan(expanded)   = {bell["tan_expanded"]!r}')
print(f'sqrt(2) - 1     = {math.sqrt(2)-1!r}')
print(f'  residual      : {abs(bell["tan_expanded"] - (math.sqrt(2)-1)):.3e}')
print()
print(f'bell_never_closes         = {bell["bell_never_closes"]!r}')
print(f'bell_never_reaches_shell  = {bell["bell_never_reaches_shell"]!r}')
print(f'GAP enforces both         = {bell["GAP_enforces_both"]!r}')
```

```python
ca = bell['contracted_angle_deg']; ea = bell['expanded_angle_deg']
fig, ax = plt.subplots(figsize=(5.6, 4.4), subplot_kw={'projection': 'polar'})
for ang, col, lab in [(ca, '#b04a3f', f'contracted {ca:.3f} deg (tan = d*)'),
                      (ea, '#3f7fb0', f'expanded {ea:.3f} deg (pi/8)')]:
    for s in (+1, -1):
        ax.plot([0, math.radians(s*ang)], [0, 1], color=col, lw=2)
    ax.plot([], [], color=col, lw=2, label=lab)
ax.set_ylim(0, 1.1); ax.set_yticks([])
ax.set_thetamin(-40); ax.set_thetamax(40)
ax.set_title('the bell swing', fontsize=10)
ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.32), fontsize=8)
plt.tight_layout(); plt.show()
print(f'swing = {bell["swing_deg"]!r} degrees')
```

## The Null Operator: e0 as the M87* spin axis

`e0 = 1` is the multiplicative identity — The Null Operator. It never
participates in zero-divisor crossings, which is what makes it usable as the
tower's reference axis. `m87_axis()` extends it into physical space.

```python
axis = tp.m87_axis()
for k, v in axis.items():
    print(f'  {k:>20}: {v!r}' if isinstance(v, float) else f'  {k:>20}: {v}')
```
