The Cayley-Dickson tower has **two** fixed points:

- `The Unit` — the k -> -inf limit, V(0) = 1, dimension 0. Trivial, pre-arithmetic.
- `T_256` — dimension 256, 255 imaginary units. All boundary, no interior.

The gap between them is the mass gap.

## The angular quantum halves at every doubling

180 degrees at C, 90 at H, 45 at O, 22.5 at S — where the first zero divisors
appear — and on down to 1.40625 degrees at T_256. Below that spacing the 256
basis elements are indistinguishable from one another, and all roots collapse
back to 1: **The Unit.**

```python
seq = fp.angular_quantum_sequence()
print(f'{"k":>2} {"algebra":<8} {"dim":>5} {"quantum(deg)":>13} {"Fano":>5} {"sigma":>8}')
print('-' * 50)
for row in seq:
    print(f'{row["k"]:>2} {str(row.get("algebra","")):<8} {row.get("dim",0):>5} '
          f'{row.get("quantum_deg", float("nan")):>13.4f} '
          f'{row.get("fano", 0):>5} {row.get("sigma", float("nan")):>8.3f}')
```

## The Transformer: V(n), the n-ball volume

V(0) = 1 exactly — The Unit. It rises to a peak near n* = 5.26, then collapses.
By n = 256 the volume is ~1e-152: the object is all surface.

```python
peak = fp.v_nball_peak()
print('peak:', peak)
print()
for n in [0, 1, 2, 4, 5, 8, 16, 32, 64, 128, 256]:
    print(f'  V({n:>3}) = {fp.v_nball(n):.6e}')

ns = np.linspace(0.01, 40, 400)
vs = [fp.v_nball(float(n)) for n in ns]
fig, ax = plt.subplots(figsize=(7.5, 3.4))
ax.plot(ns, vs, color='#3f7fb0')
ax.axvline(fp.N_STAR, color='#b04a3f', ls='--', lw=1,
           label=f'n* = {fp.N_STAR:.4f}')
ax.axhline(1.0, color='#999', ls=':', lw=1, label='V(0) = 1 (The Unit)')
ax.axvline(16, color='#2f8f5f', ls=':', lw=1, label='n = 16 (sedenion)')
ax.set_xlabel('n'); ax.set_ylabel('V(n)')
ax.set_title('the n-ball transformer: peak differentiation then collapse')
ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```
