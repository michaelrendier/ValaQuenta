## THE ANGLE = pi/8 = 22.5 degrees

`tan(pi/8) = sqrt(2) - 1`, reproduced to machine epsilon. Sixteen of them make
one revolution — which is exactly the 16 dimensions of the sedenion.

Its job: before rotation the red and blue current sectors *interleave*. After
rotating by pi/8 they **coincide**.

```python
angle = math.pi / 8
print(f'THE ANGLE      = pi/8 = {math.degrees(angle)!r} degrees')
print(f'tan(pi/8)      = {math.tan(angle)!r}')
print(f'sqrt(2) - 1    = {math.sqrt(2)-1!r}')
print(f'residual       = {abs(math.tan(angle)-(math.sqrt(2)-1)):.3e}')
print(f'16 * pi/8      = {math.degrees(16*angle)!r} degrees = one revolution')
print()
red_before  = [0.0, 90.0, 180.0, 270.0]
blue_before = [45.0, 135.0, 225.0, 315.0]
red_after   = [(a + 22.5) % 360 for a in red_before]
blue_after  = [(a - 22.5) % 360 for a in blue_before]
print(f'J_red  before: {red_before}')
print(f'J_blue before: {blue_before}   -> interleaved')
print(f'J_red  after : {sorted(red_after)}')
print(f'J_blue after : {sorted(blue_after)}   -> coincide')
print(f'coincide     : {sorted(red_after) == sorted(blue_after)}')
```

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(9, 4.4),
                             subplot_kw={'projection': 'polar'})
for ax, (r, b, title) in zip(
        (a1, a2),
        [(red_before, blue_before, 'before: interleaved'),
         (red_after, blue_after, 'after pi/8: coincident')]):
    ax.plot(np.radians(r), [1]*4, 'o', ms=11, color='#b04a3f', label='J_red')
    ax.plot(np.radians(b), [1]*4, 'x', ms=13, mew=2.5, color='#3f7fb0',
            label='J_blue')
    ax.set_ylim(0, 1.35); ax.set_yticks([]); ax.set_title(title, fontsize=10)
a2.legend(loc='upper right', bbox_to_anchor=(1.25, 1.12), fontsize=8)
plt.tight_layout(); plt.show()
```
