## Sampling by phase: why a lattice makes spokes
Equal steps of theta put every sample on 2*16 fixed angles of the flat point. A golden-ratio jitter removes it.

```python
t0 = M.sample_by_phase(10.0, 400, 16); t1 = M.sample_by_phase(10.0, 400, 16, jitter=True)
print('distinct angles, lattice:', len(np.unique(np.round(np.mod(-M.theta(t0), np.pi), 6))), ' jittered:', len(np.unique(np.round(np.mod(-M.theta(t1), np.pi), 6))))
```
