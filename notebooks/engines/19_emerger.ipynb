# 19 — The Emerger

**Sedenion Bracketing & Firing Order.** A dynamic permutative bracketer over the imaginary part of a Cayley-Dickson algebra. `e_0` (real) is the fixed anchor — the tilt to the *i* axis — never bracketed. Each imaginary group + the anchor spans a domain (C / H / O / FRAGMENT). The order the groups are approached (the firing order) is load-bearing.

The ascent-dual of Generational Lineage: descent = what built this (writing); ascent = what emerges, in what order (reading — spectroscopy, factoral decomposition).


```python
import sys, os
sys.path.insert(0, os.path.abspath('../../..'))
from ValaQuenta.modules.emerger import EmergerModule, emerge, firing_order, verify, scale_partitions, legal_orders, lineage_report, domain_of
m = EmergerModule()
print(m.display_name, m.version, '| floor:', m.confidence_floor)
```

## Emergence readout — `e1 + e10` (a zero divisor, on the ZD equator)


```python
r = emerge('e1+e10', mode='sigma_rb')
fo = r['firing_order']
print('Sigma_tilt = %.4f   precession step %d/12   entry bracket #%d' %
      (fo['Sigma_tilt'], fo['precession_step_of_12'], fo['entry_bracket_index']))
print('canonical :', ' -> '.join(fo['canonical']))
print('phased    :', ' -> '.join(fo['sigma_rb_phased']), ' legal:', fo['phased_is_legal'])
print()
for s in r['steps']:
    extra = {k:v for k,v in s.items() if k not in
             ('bracket','tier','descends_from','emerges','conditioned_on','groups')}
    print(f"[{s['tier'][:2]}] {s['bracket']:11s} {extra}")
```
