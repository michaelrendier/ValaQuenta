## The Holcus prediction

`K_core(M) = 24 / L_dS^4 = 24 / r_s^4 = (3/2) c^8 / (G^4 M^4)` — the de Sitter Kretschmann scalar at `L_dS = r_s`. Finite for all `M>0`, scales as `M^-4`, sub-Planckian for every `M > (3/2)^(1/4) m_Pl`.

```python
Msun = M.M_SUN
for label, Mkg in [('kugelblitz 1e12 kg', 1e12), ('stellar 10 Msun', 10*Msun),
                   ('IMBH 1e4 Msun', 1e4*Msun), ('SMBH 1e9 Msun', 1e9*Msun)]:
    r = E.run('kretschmann_core', {'M_kg': Mkg})['result']
    print(f"{label:20s}  K_core={r['K_core_m^-4']:.3e} m^-4   "
          f"K/K_Planck={r['K_core_over_K_Planck']:.2e}   "
          f"closed-form matches={r['closed_form_matches']}")
print()
print('Planck-mass crossover:', M.planck_mass_crossover())
```
