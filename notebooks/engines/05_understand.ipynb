# Engine: Understand — Read, Listen, Ponder, Calculate, Understand

**File:** `ValaQuenta/understand.py`
**Wiki:** [wiki/understand.md](../../wiki/understand.md)

Five operations, one tool. No inference, no training data, no GPU.

```
Read       surface form  -> candidate prime
Listen     acoustic      -> formant prime
Ponder     H = xp        -> conserved prime
Calculate  operations    -> projections
Understand capacitor     -> DC prime
```

## sigma = 1/2 on every input, not just the convenient one

````python
probes = ['tree', 'the prime', 'a', 'Ainulindale',
          'zero divisors in the sedenion algebra',
          '', '42', 'why is the mass gap 1 over root 2000']

print(f'{"input":>42} | {"sigma":>20} | {"|s-0.5|":>9} | faces')
print('-' * 90)
worst = 0.0
for text in probes:
    ww = U.process(text)
    s = ww.prime.real
    worst = max(worst, abs(s - 0.5))
    print(f'{text[:42]:>42} | {s!r:>20} | {abs(s-0.5):9.2e} | {ww.faces()}')
print()
print(f'worst deviation from 1/2 across all inputs: {worst:.3e}')
````

The empty string is included on purpose. An engine that derives sigma
= 1/2 from *nothing* is either deriving it from the geometry, or the value is
hard-wired. The wiki records this as derived; this cell is where a reader can
check that claim against a degenerate input.

````python
# The five operations, run one at a time rather than through process().
domain = U.describe('mathematics, the Riemann zeros, the critical line')
print('domain     :', domain)
print('  span         =', domain.span)
print('  hawking_temp =', repr(domain.hawking_temperature()))

w1 = U.read('tree', domain=domain)
print('\nread       :', w1)
w2 = U.ponder(w1, t=1.0)
print('ponder     :', w2)
w3 = U.calculate(w2)
print('calculate  :', w3)
w4 = U.understand(w3)
print('understand :', w4)
print(f'\nfinal dc = {w4.dc!r}   sigma = {w4.prime.real!r}')
````
