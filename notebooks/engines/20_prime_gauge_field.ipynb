## Two candidate connections

`A = grad(log|Gamma|)` — provably flat for **any** holomorphic scalar
(Poincare lemma, `d(df)=0` identically). This generalizes the earlier
Schwarzian-zero finding: it was never a special property of `Gamma`.

`A = (Re Gamma, Im Gamma)` — `Gamma` read directly as an `R^2`-valued
1-form, *not* a gradient of anything. This one can carry curvature.

```python
for s in [0.3+0.7j, 2-1j, -0.5+2j, 0.5+0.1j, 3+0j, -1+2j]:
    r = report(s)
    print(f"s={s!s:>14}  Gamma={r['Gamma']:.4f}  "
          f"curvature={r['curvature']:+.6f}  "
          f"log_potential_curvature={r['log_potential_curvature']:.2e}")
```

## Closed form vs. finite difference — the honesty check

`F(s) = 2*Im(dGamma/ds)` via Cauchy-Riemann. Verified against a
finite-difference numeric derivative before being trusted anywhere
else in this engine.

```python
import numpy as np
errs = []
for s in [0.3+0.7j, 2-1j, -0.5+2j, 0.5+0.1j, -3+5j, 1.5-2j]:
    c, cn = curvature(s), curvature_numeric(s)
    errs.append(abs(c-cn))
    print(f"s={s!s:>14}  closed={c:+.8f}  numeric={cn:+.8f}  |diff|={abs(c-cn):.2e}")
print(f"\nmax error: {max(errs):.2e}")
```
