```
L_NN = (2/π) ∮ [L_kin + L_mat + (1/φ)·L_bias + L_coup] r dr dθ

Running coupling:
  α_NN(r) = g² / (4π · ħ_NN · ln(1/r))
```

All arithmetic is radian-primary. `fractions.Fraction` throughout; float only at output boundary.

````python
def _show(obj, indent=0):
    """Print whatever the engine actually returned, without assuming a schema."""
    pad = ' ' * indent
    if isinstance(obj, dict):
        width = max((len(str(k)) for k in obj), default=0)
        for k, v in obj.items():
            if isinstance(v, (dict, list, tuple)) and v:
                print(f'{pad}{str(k):>{width}} :')
                _show(v, indent + width + 3)
            else:
                print(f'{pad}{str(k):>{width}} : {v!r}')
    elif isinstance(obj, (list, tuple)):
        for j, v in enumerate(obj[:12]):
            if isinstance(v, (dict, list, tuple)) and v:
                print(f'{pad}[{j}]')
                _show(v, indent + 3)
            else:
                print(f'{pad}[{j}] {v!r}')
        if len(obj) > 12:
            print(f'{pad}... +{len(obj)-12} more')
    else:
        print(f'{pad}{obj!r}')
````

## Running Coupling α_NN(r)

````python
# Regenerated 2026-07-28: maths.running_coupling() is now alpha_nn_from_r(),
# with the argument order (g, hbar_nn, r).
ALG_NAMES = {0: 'R', 1: 'C', 2: 'H', 3: 'O'}
print(f'{"r":>6} {"alpha_NN(r)":>16}   algebra')
for r in [0.01, 0.1, 0.5, 0.8, 0.9, 0.99]:
    alpha = lg.alpha_nn_from_r(1.0, 1.0, r)
    stratum = 0 if r > 0.8 else (1 if r > 0.5 else (2 if r > 0.1 else 3))
    print(f'{r:>6.2f} {alpha:>16.8f}   ({ALG_NAMES[stratum]})')
````

## Full Lagrangian Density

````python
# Regenerated 2026-07-28: maths.lagrangian_density() is now polar_lagrangian(),
# which returns the term breakdown as a dict rather than a single scalar.
# The dict mixes floats with descriptive strings, so format per value type.
ALG_NAMES = {0: 'R', 1: 'C', 2: 'H', 3: 'O'}
for alg in [lg.ALG_R, lg.ALG_C, lg.ALG_H, lg.ALG_O]:
    fs_alg = lg.FieldState(psi=psi, A=A, beta=beta, algebra=alg, layer=alg + 1)
    L = lg.polar_lagrangian(fs_alg)
    print(f'--- algebra {ALG_NAMES.get(alg, alg)}')
    for k, v in L.items():
        print(f'    {k:>14} : {v:+.6e}' if isinstance(v, (int, float))
              else f'    {k:>14} : {v!r}')
print()
print(f'TWO_OVER_PI = {float(lg.TWO_OVER_PI)!r}')
````
