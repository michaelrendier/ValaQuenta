## Part 4 — The Niemeier gap (the theorem under the 'N-shape' claim)

**Theorem.** An ADE root system of rank 24 whose components share one Coxeter number h has h mod 16 outside {1, 11, 15}.

*Proof.* D_n and E_n have even h, so an odd h forces every component to be A_(h−1), and then (h−1) divides 24: the odd h are 3, 5, 7, 9, 13, 25. ∎ The code generates the 23 root systems from the defining condition and checks each step.

```python
t = hg.niemeier_gap_theorem()
print('root systems:', t['count'])
for s in t['root_systems']:
    print(f"  h={s['h']:2d}  {s['label']}")
print('h values          :', t['h_values'])
print('residues mod 16   :', t['residues_reached'])
print('gap               :', t['gap'])
print('proof steps       :', t['proof'])
print('theorem holds     :', t['theorem_holds'])
```
