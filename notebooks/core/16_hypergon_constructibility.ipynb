## Reading the result honestly

Close real factor pairs: p=69.1%, q=76.4% nilpotent (well above 50% baseline —
this is the number Phase 22 reported, reproduced here).

Far-apart real factor pairs: p=47.4%, q=45.4% — at or below baseline, the
elevation is gone.

Random pairs (NOT factors of any common N): p=50.5%, q=56.7% — comparable to
or higher than the far-apart REAL factor pairs.

If the mechanism actually tracked factoring relationships, far-apart factor
pairs should beat the random control. They don't. The close-pair elevation
is consistent with a magnitude artifact of how small numbers happen to map
through the Hyperwebster address into T32/GF(2), not with genuine
factoring-relevant structure.
