## Step 1 — Kaiser-Squires: real shear -> convergence (exact, no fitting)

```python
# Small synthetic shear field -- for demonstrating the MECHANISM only.
# The real science result (Bullet Cluster JWST shear) is cited, not re-run, below --
# this cell exists to show the pipeline and the k=0 handling on inspectable numbers.
rng = np.random.default_rng(42)
gamma1 = rng.normal(0, 0.05, size=(16, 16))
gamma2 = rng.normal(0, 0.05, size=(16, 16))

kappa = lio.kaiser_squires_kappa(gamma1, gamma2)
print(f'kappa: mean={kappa.mean():.6f}  std={kappa.std():.6f}  shape={kappa.shape}')
```

## Step 2 — the k=0 boundary point, made explicit

This is the reusable part: not the lensing physics, the *convention* at the transform's own singular point.

```python
k1 = np.fft.fftfreq(16) * 2 * np.pi
k2 = np.fft.fftfreq(16) * 2 * np.pi
K1, K2 = np.meshgrid(k1, k2)
k2mag = K1**2 + K2**2
print(f'k=0 mode of |k|^2 is exactly {k2mag[0,0]}  -- the direct formula 1/|k|^2 diverges here.')
print('kaiser_squires_kappa does not divide by this: it substitutes k2mag[0,0]=1.0 to')
print('avoid the crash, THEN explicitly sets kappa_hat[0,0]=0.0 -- an assigned value,')
print('with a stated reason (the mean is not observable from shear alone), not a fit')
print('and not a silent omission. This explicit-assignment-at-the-singular-point is')
print('the general template L_(I|O) provides for crossing a boundary once (I|O)_RB')
print('has defined where that boundary is.')
```

## Step 3 — Poisson solve, deflection, lens equation (all exact, chained)

```python
psi = lio.lensing_potential(kappa)
alpha1, alpha2 = lio.deflection_field(psi, pixel_scale_arcsec=1.0)
trace = lio.trace_photon(np.zeros((16,16)), np.zeros((16,16)), alpha1, alpha2)
deficit = lio.l_io_deficit(psi)

print(f"psi:    mean={psi.mean():.6f}  std={psi.std():.6f}")
print(f"deflection_mag: mean={trace['deflection_mag'].mean():.6f}")
print(f"L_(I|O) - L = -psi(theta): {deficit}")
```
