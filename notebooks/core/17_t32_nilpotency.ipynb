## The GF(2) square law and the detector (0.110)

Every word satisfies x² ∈ {0, e₀}: nilpotent (the radical) or involutory unit. The detector Δ(w) = w·𝟏 vanishes exactly on the nilpotents. 𝟏 is **not** a global annihilator (corrected lemma, 2026-08-22).

```python
for d in (2, 4, 8):
    print(t32.radical_census(d))
import random
random.seed(1)
ws = [random.getrandbits(32) for _ in range(20000)]
print('dim 32, 20000 random words: detector mismatches =', sum((t32.trace_laplacian(w) == 0) != (t32.is_nilpotent(w) or w == 0) for w in ws), '; classes =', sorted({t32.square_class(w) for w in ws}))
```

## The control the page lacked (0.120)

A random word is nilpotent half the time; so are the Hyperwebster words of the integers from three digits up, primes and composites alike. The prime excess is a two-digit-range effect. The chance of 5 of 5 gap primes among random primes of the same size is the honest baseline.

```python
for lo, hi in ((10, 100), (100, 1000), (1000, 10000)):
    w = t32.nilpotency_window(lo, hi)
    print(lo, hi, {k: (v['nilpotent'], v['n'], round(v['rate'], 3)) for k, v in w.items() if isinstance(v, dict)})
print(t32.matched_prime_control([11, 17, 31, 47, 59]))
```
