## What this module is, and isn't

**Is:** the one place this specific mechanism should be maintained — imported by
two engines across two repos now, rather than copy-pasted, closing off the exact
class of bug (silent divergence between "copies" of the same function) that cost
real debugging time building `hypergon_constructibility`.

**Isn't:** validated as a factoring or Niemeier-gap-detection mechanism. The
factoring use (`hypergon_constructibility`) was tested against a proper control
and found to be an artifact. The Niemeier-gap correlation
(`fermat_monster_engine.t32_nilpotency_check`) is a real raw number on a small
sample, not yet control-tested the same way — don't treat it as more settled
than that.
